```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------------------------------------------------
# NOTE ABOUT AI USAGE
# -----------------------------------------------------------------------
# The graphs and plotting code were fine-tuned with the help of AI.
# AI was used to improve the visual styling, organize the plots into groups,
# automate the creation of multiple histograms and boxplots, and make the
# saved graph images clearer and more professional for the report.
# The dataset, chosen features, and interpretation are still part of my own project work.

df = pd.read_csv(r'C:\Users\Velimir\iCloudDrive\Documents\Python for AI\Project\Large set\ufc_processed.csv')

# -----------------------------------------------------------------------
# Column groups
# -----------------------------------------------------------------------
physical  = ['r_age', 'b_age', 'r_height', 'b_height', 'r_weight', 'b_weight']
record    = ['r_wins_total', 'r_losses_total', 'b_wins_total', 'b_losses_total']
striking  = ['r_SLpM_total', 'r_SApM_total', 'r_sig_str_acc_total', 'r_str_def_total',
             'b_SLpM_total', 'b_SApM_total', 'b_sig_str_acc_total', 'b_str_def_total']
grappling = ['r_td_avg', 'r_td_acc_total', 'r_td_def_total', 'r_sub_avg',
             'b_td_avg', 'b_td_acc_total', 'b_td_def_total', 'b_sub_avg']
diffs     = ['wins_total_diff', 'age_diff', 'height_diff', 'weight_diff',
             'SLpM_total_diff', 'SApM_total_diff', 'sig_str_acc_total_diff',
             'td_acc_total_diff', 'str_def_total_diff', 'td_def_total_diff']

groups = {
    'Physical Attributes': physical,
    'Fight Record':        record,
    'Striking Stats':      striking,
    'Grappling Stats':     grappling,
    'Diff Columns':        diffs,
}

# -----------------------------------------------------------------------
# Style
# -----------------------------------------------------------------------
red   = '#e63946'
blue  = '#457b9d'
teal  = '#a8dadc'
bg    = '#1a1a2e'
panel = '#16213e'
text  = '#eaeaea'
grid  = '#2a2a4a'

def col_color(col):
    if col.startswith('r_'):
        return red
    elif col.startswith('b_'):
        return blue
    return teal

def make_plots(plot_type):
    for group_name, cols in groups.items():
        n     = len(cols)
        ncols = 4
        nrows = int(np.ceil(n / ncols))

        fig, axes = plt.subplots(nrows, ncols, figsize=(18, nrows * 3.2))
        fig.patch.set_facecolor(bg)
        title = ('Distributions' if plot_type == 'dist' else 'Outliers')
        fig.suptitle(f'{title} — {group_name}', color=text, fontsize=15,
                     fontweight='bold', y=1.01)
        axes = axes.flatten()

        for i, col in enumerate(cols):
            ax    = axes[i]
            color = col_color(col)
            data  = df[col].dropna()
            ax.set_facecolor(panel)

            if plot_type == 'dist':
                ax.hist(data, bins=40, color=color, alpha=0.85, edgecolor='none')
            else:
                ax.boxplot(data, patch_artist=True, vert=True,
                           medianprops=dict(color='white', linewidth=2),
                           flierprops=dict(marker='o', markerfacecolor=color,
                                           markersize=2, alpha=0.4),
                           whiskerprops=dict(color=text),
                           capprops=dict(color=text),
                           boxprops=dict(facecolor=color, alpha=0.7))
                ax.set_xticks([])

            ax.set_title(col, color=text, fontsize=9, pad=4)
            ax.tick_params(colors=text, labelsize=7)
            for spine in ax.spines.values():
                spine.set_edgecolor(grid)
            ax.yaxis.grid(True, color=grid, linewidth=0.5)
            ax.set_axisbelow(True)

        for j in range(i + 1, len(axes)):
            axes[j].set_visible(False)

        plt.tight_layout()
        safe = group_name.replace(' ', '_')
        filename = f'{plot_type}_{safe}.png'
        plt.savefig(filename, dpi=150, bbox_inches='tight', facecolor=bg)
        plt.close()
        print(f'Saved {filename}')

make_plots('dist')
make_plots('box')
```
